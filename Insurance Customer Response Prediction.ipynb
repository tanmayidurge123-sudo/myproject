# <b><span style='color:#1F4E79'>Step 5.4 |</span><span style='color:#1F4E79'> Train-Test Split </span></b>


<div style="border-radius:10px; padding:15px; background-color:#1F4E79; font-size:115%; text-align:left; color:white;">

<p>
The dataset was divided into training and testing sets to evaluate model performance on unseen data. 
80% of the data was used for training and 20% for testing to ensure reliable model evaluation.

</div>

<div style="border-radius:10px; padding:15px; background-color:#1F4E79; font-size:115%; text-align:left; color:white;">

<p>
The target variable showed class imbalance, with a higher number of non-interested customers compared to interested customers. 
SMOTE was applied to balance the classes and improve the model’s ability to predict minority class responses effectively.
</p>

</div>

```python
print("\n Applying SMOTE")

print("Before SMOTE:")
print(y_train.value_counts())

smote = SMOTE(random_state=42)
X_train, y_train = smote.fit_resample(X_train, y_train)

print("\nAfter SMOTE:")
print(pd.Series(y_train).value_counts())

```

```python
sns.countplot(x=y_train)
plt.title("Balanced Data After SMOTE")
plt.xlabel("Response")
plt.ylabel("Count")
plt.show()
```

<h3>Inference:</h3>

<ul>

<li>SMOTE successfully balanced the target classes in the training dataset.</li>

<li>This helps reduce bias towards the majority class and improves prediction performance.</li>

</ul>
